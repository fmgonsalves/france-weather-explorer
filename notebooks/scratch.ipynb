```python
import pyarrow.parquet as pq
from pathlib import Path
```

```python
def print_metadata(path: Path):
    metadata = pq.read_metadata(path)
    data = pq.read_table(path)

    print(metadata)
    print(data)
```

```python
filepath = Path("../data/meteo_france/metadata/department=44/station_inventory.parquet")

print_metadata(filepath)
```
